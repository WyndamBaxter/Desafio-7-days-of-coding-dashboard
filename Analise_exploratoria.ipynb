DIA 01 - Importando dados

```python
#IMPORTANDO BIBLIOTECAS
import pandas as pd
import plotly_express as px
import os

#IMPORTANDO DATASETS
caminho_datasets = 'datasets'
lista_de_diretorios = os.listdir(caminho_datasets)

dfs = []

for arquivo in lista_de_diretorios:
    if arquivo.endswith('.csv'):
        caminho_completo = os.path.join(caminho_datasets, arquivo)
        df = pd.read_csv(caminho_completo)
        dfs.append(df)
    
df_emprestimos_biblioteca = pd.concat(dfs, ignore_index= True)
```

```python
df_emprestimos_biblioteca
```

```python
df_emprestimos_biblioteca.value_counts()
```

```python
df_emprestimos_biblioteca = df_emprestimos_biblioteca.drop_duplicates()
df_emprestimos_biblioteca
```

Importando banco de dados .parquet

```python
dados_exemplares = pd.read_parquet('datasets/dados_exemplares.parquet')
dados_exemplares
```

Unindo os dois datasets

```python
df_emprestimos_completo = pd.merge(df_emprestimos_biblioteca, dados_exemplares, on= 'codigo_barras')
df_emprestimos_completo
```

```python
df_emprestimos_completo.info()
```

Quantidade de exemplares emprestados por ano

Primeiro converter  coluna de data para uma data válida

```python
df_emprestimos_completo['data_emprestimo'] = pd.to_datetime(df_emprestimos_completo['data_emprestimo'])
```

```python
emprestimo_ano = df_emprestimos_completo.groupby(df_emprestimos_completo['data_emprestimo'].dt.year)['id_emprestimo'].size()
emprestimo_ano.index.name ='ano'
emprestimo_ano.name = 'quantidade'
emprestimo_ano = emprestimo_ano.reset_index()
```

Impressão gráfico de linha:


```python
fig = px.line(
    emprestimo_ano,
    x = 'ano',
    y = 'quantidade',
    labels= {
        'ano': 'Ano',
        'quantidade':'Empréstimos'
        },
    title = 'Quantidade de exemplares emprestados do SISBI ao longo dos anos',
    subtitle= '2010 a 2020',
    width = 1000,
    height = 500
)
```

```python
def ajusta_layout_grafico(fig):
    
    fig.update_layout(
        title_font = dict(size=20, family='Arial', color='black'),
    )

    fig.update_xaxes(
        title_font = dict(size=15, family='Arial'),
        tickfont = dict(size=15, family='Arial'),
        side = 'bottom' # 'bottom' é o padrão, mas é bom para garantir
    )

    fig.update_yaxes(
        title_font = dict(size=15, family='Arial'),
        tickfont = dict(size=15, family='Arial')
    )
```

```python
fig.show()
ajusta_layout_grafico(fig)
```

1 - O número de empréstimos começou a diminuir em 2013, tendo queda vertiginosa em 2019.

2 - O ano de 2020 não pode ser valiado igual, pois possui poucos dados.

Vamos analisar ao longo do dia.

```python
emprestimos_hora = df_emprestimos_completo.groupby(df_emprestimos_completo['data_emprestimo'].dt.hour)['id_emprestimo'].size()
emprestimos_hora.index.name = 'hora'
emprestimos_hora.name = 'quantidade'
emprestimos_hora = emprestimos_hora.reset_index()
emprestimos_hora
```

Organizando valores e plotando gráfico

```python
emprestimos_hora = emprestimos_hora.sort_values(ascending=True, by='quantidade')
```

```python
fig = px.bar(
    emprestimos_hora,
    x = 'hora',
    y = 'quantidade',
    color='quantidade',
    color_continuous_scale= 'reds',
    labels={
        'hora': 'Horário (24h)',
        'quantidade': 'Quantidade de empréstimos'
    },
    title= 'Quantidade de empréstimos do SISBI ao longo do dia',
    width= 1300,
    height= 500
)
```

```python
fig.show()
ajusta_layout_grafico(fig)
```

Através do gráfico é possível identificar que os períodos das 10 h e 16 h são os de maior pico no atendimento de empréstimos.

Períodos do começo da manhã (entre 7h e 9) e da noite (entre 19h e 22) podem ser os melhores horários para que se realizem as demais atividades além do atendimento.

Variável categórica? Mas o que é isso, Francisco?

Bom, um conjunto de dados é descrito pelas suas características que variam de acordo com cada um, por exemplo: peso, idade, quantidade de páginas, notas, nome da biblioteca, número de empréstimos, número de filhos, etc.

Elas podem ser classificadas em dois grandes grupos que se subdividem em mais dois outros grupos cada.

Quantitativas: são características numéricas que podemos enumerar:


Discreta: possuem apenas valores finitos enumeráveis. Exemplo: número de empréstimos, número de filhos, número de páginas.

Contínua: possuem valores dentro de um intervalo de números. Exemplo: altura, peso, notas.

Categóricas (ou qualitativas): são características não-numéricas para as quais podemos definir categorias.


Nominal: possuem categorias não ordenadas naturalmente. Exemplo: nomes, cores, tema do livro.

Ordinal: são categorias onde está implícito um tipo de ordenamento ou hierarquia. Exemplo: classe social, nível de escolaridade.
