DIA 01 - Importando dados

```python
#IMPORTANDO BIBLIOTECAS
import pandas as pd
import plotly_express as px
import os

#IMPORTANDO DATASETS
caminho_datasets = 'datasets'
lista_de_diretorios = os.listdir(caminho_datasets)

dfs = []

for arquivo in lista_de_diretorios:
    if arquivo.endswith('.csv'):
        caminho_completo = os.path.join(caminho_datasets, arquivo)
        df = pd.read_csv(caminho_completo)
        dfs.append(df)
    
df_emprestimos_biblioteca = pd.concat(dfs, ignore_index= True)
```

```python
df_emprestimos_biblioteca
```

```python
df_emprestimos_biblioteca.value_counts()
```

```python
df_emprestimos_biblioteca = df_emprestimos_biblioteca.drop_duplicates()
df_emprestimos_biblioteca
```

Importando banco de dados .parquet

```python
dados_exemplares = pd.read_parquet('datasets/dados_exemplares.parquet')
dados_exemplares
```

Unindo os dois datasets

```python
df_emprestimos_completo = pd.merge(df_emprestimos_biblioteca, dados_exemplares, on= 'codigo_barras')
df_emprestimos_completo
```

```python
df_emprestimos_completo.info()
```

DIA 02 - Primeiras manipulações

Criação da coluna 'Classificação CDU' baseada na coluna localização.

```python
df_emprestimos_completo['classe_CDU'] = pd.cut(df_emprestimos_completo['localizacao'], bins= [0, 99, 199, 299,399,499,599,699,799,899,999], labels= ['Generalidades. Ciência e conhecimento', 'Filosofia e psicologia', 'Religião', 'Ciências Sociais', 'Classe vaga', 'Matemática e ciências naturais', 'Ciências aplicadas', 'Belas artes', 'Linguagem. Língua. Linguística', 'Geografia. Biografia. História'] )
df_emprestimos_completo
```

Excluindo a coluna 'registro_sistema'

```python
df_emprestimos_completo.drop(columns='registro_sistema', inplace=True)
df_emprestimos_completo
```

Verificando o tipo da coluna 'matricula_ou_siape' e convertendo-a para string 

```python
df_emprestimos_completo.info()
```

```python
df_emprestimos_completo['matricula_ou_siape'] = df_emprestimos_completo['matricula_ou_siape'].astype('string')
df_emprestimos_completo.info()
```
